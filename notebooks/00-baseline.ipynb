```python
import pandas as pd

data= pd.read_parquet('../data/green_tripdata_2025-01.parquet')


print(data.shape)
```

```python
# Missing values
data.isnull().sum()
```

```python
# Duplicate rows
data.duplicated().sum()
```

```python
# 1. Drop exact duplicate rows
data = data.drop_duplicates()

# 2. Drop rows with missing critical fields
data = data.dropna(subset=['lpep_pickup_datetime', 'lpep_dropoff_datetime', 'trip_distance', 'fare_amount'])

# 3. Filter pickup times to the expected month (removes stray rows from adjacent months)
data = data[(data['lpep_pickup_datetime'] >= '2025-01-01') & (data['lpep_pickup_datetime'] < '2025-02-01')]

# 4. Remove trips with non-positive distance or fare
data = data[data['trip_distance'] > 0]
data = data[data['fare_amount'] > 0]

# 5. Remove trips with unreasonable duration (e.g. negative or absurdly long)
data['trip_duration_min'] = (data['lpep_dropoff_datetime'] - data['lpep_pickup_datetime']).dt.total_seconds() / 60
data = data[(data['trip_duration_min'] > 0) & (data['trip_duration_min'] < 180)]  # cap at 3 hours

# 6. Remove passenger_count outliers (0 passengers is common bad data)
data = data[data['passenger_count'] > 0]

# 7. Cap extreme fare/total_amount outliers (optional, adjust threshold as needed)
data = data[data['total_amount'] < data['total_amount'].quantile(0.99)]

# 8. Reset index after all the filtering
data = data.reset_index(drop=True)

print(data.shape)
data.isnull().sum()
```

```python
data = data.dropna(subset=['trip_type'])
data = data.drop(columns=['ehail_fee'])
```

```python
import matplotlib.pyplot as plt

fig, axes = plt.subplots(2, 2, figsize=(12, 8))
data['trip_distance'].hist(bins=50, ax=axes[0,0])
axes[0,0].set_title('Trip Distance')

data['trip_duration_min'].hist(bins=50, ax=axes[0,1])
axes[0,1].set_title('Trip Duration (min)')

data['fare_amount'].hist(bins=50, ax=axes[1,0])
axes[1,0].set_title('Fare Amount')

data['total_amount'].hist(bins=50, ax=axes[1,1])
axes[1,1].set_title('Total Amount')

plt.tight_layout()
plt.show()
```

```python
import seaborn as sns

numeric_cols = ['trip_distance', 'trip_duration_min', 'fare_amount', 'tip_amount', 'total_amount', 'passenger_count']
corr = data[numeric_cols].corr()

plt.figure(figsize=(8, 6))
sns.heatmap(corr, annot=True, cmap='coolwarm', fmt='.2f')
plt.show()
```

```python
data['pickup_hour'] = data['lpep_pickup_datetime'].dt.hour
data['pickup_dayofweek'] = data['lpep_pickup_datetime'].dt.dayofweek  # 0=Monday
data['pickup_month'] = data['lpep_pickup_datetime'].dt.month
data['is_weekend'] = data['pickup_dayofweek'].isin([5, 6]).astype(int)
```

```python
data.groupby('pickup_hour').size().plot(kind='bar', figsize=(10, 4), title='Trips by Pickup Hour')
plt.show()
```

```python
data['PU_DO'] = data['PULocationID'].astype(str) + '_' + data['DOLocationID'].astype(str)
```

```python
from sklearn.model_selection import train_test_split

features = ['PU_DO', 'trip_distance']
target = 'trip_duration_min'

train, val = train_test_split(data, test_size=0.2, random_state=42)

print(train.shape, val.shape)
```

```python
from sklearn.feature_extraction import DictVectorizer

dv = DictVectorizer()

train_dicts = train[features].to_dict(orient='records')
val_dicts = val[features].to_dict(orient='records')

X_train = dv.fit_transform(train_dicts)
X_val = dv.transform(val_dicts)

y_train = train[target].values
y_val = val[target].values
```

```python
from sklearn.linear_model import LinearRegression

model = LinearRegression()
model.fit(X_train, y_train)
```

```python
from sklearn.metrics import mean_squared_error, mean_absolute_error
import numpy as np

y_pred = model.predict(X_val)

rmse = np.sqrt(mean_squared_error(y_val, y_pred))
mae = mean_absolute_error(y_val, y_pred)

print(f'Validation RMSE: {rmse:.3f}')
print(f'Validation MAE: {mae:.3f}')
```

```python
import pickle
import os

os.makedirs('../models', exist_ok=True)

with open('../models/baseline.pkl', 'wb') as f_out:
    pickle.dump((dv, model), f_out)
```
